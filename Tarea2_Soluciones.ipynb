# PREGUNTA 01: MAP con LISTA

Dada la lista `temperaturas = [18, 21, 25, 30, 16, 22]`, convertir todas las temperaturas a Fahrenheit usando 4 formas distintas.

### 1. Usando un for loop tradicional

```python
temperaturas = [18, 21, 25, 30, 16, 22]
fahrenheit_for = []

for temp in temperaturas:
    f = (temp * 9/5) + 32
    fahrenheit_for.append(f)

print("For loop:", fahrenheit_for)
```

- Se crea una lista vacía `fahrenheit_for`.
- Se recorre cada temperatura con un `for`, se aplica la fórmula y se agrega el resultado.

### 2. Usando map() con una función normal

```python
def celsius_a_fahrenheit(c):
    return (c * 9/5) + 32

fahrenheit_map = list(map(celsius_a_fahrenheit, temperaturas))
print("map() con función:", fahrenheit_map)
```

- `celsius_a_fahrenheit` es una función normal que recibe un valor y retorna el convertido.
- `map()` aplica esa función a cada elemento de la lista y devolvemos una lista con `list()`.

### 3. Usando map() con lambda

```python
fahrenheit_lambda = list(map(lambda c: (c * 9/5) + 32, temperaturas))
print("map() con lambda:", fahrenheit_lambda)
```

- Es equivalente al caso anterior, pero la función se define inline con `lambda`.

### 4. Usando list comprehension

```python
fahrenheit_comp = [(c * 9/5) + 32 for c in temperaturas]
print("List comprehension:", fahrenheit_comp)
```

- La forma más compacta y pitónica. Hace lo mismo que el for loop en una sola línea.

# PREGUNTA 02: FILTRO con TUPLA

Dada la tupla `edades = (12, 17, 19, 21, 15, 30, 16, 40)`, filtrar únicamente las edades mayores o iguales a 18 usando 4 formas distintas.

### 1. Usando un for loop tradicional

```python
edades = (12, 17, 19, 21, 15, 30, 16, 40)
mayores_for = []

for edad in edades:
    if edad >= 18:
        mayores_for.append(edad)

print("For loop:", mayores_for)
```

- Se crea una lista vacía, se recorre la tupla y solo se agregan las edades que cumplen la condición.

### 2. Usando filter() con una función normal

```python
def es_mayor_edad(edad):
    return edad >= 18

mayores_filter = list(filter(es_mayor_edad, edades))
print("filter() con función:", mayores_filter)
```

- `es_mayor_edad` retorna `True` si la edad cumple la condición.
- `filter()` conserva solo los elementos donde la función retorna `True`.

### 3. Usando filter() con lambda

```python
mayores_lambda = list(filter(lambda edad: edad >= 18, edades))
print("filter() con lambda:", mayores_lambda)
```

- Mismo comportamiento que el anterior, con la función definida inline mediante `lambda`.

### 4. Usando list comprehension

```python
mayores_comp = [edad for edad in edades if edad >= 18]
print("List comprehension:", mayores_comp)
```

- La forma más concisa. El `if` al final filtra los elementos que no cumplen la condición.
