# 1. Eliminar registros duplicados

Una empresa registra las compras realizadas por sus clientes. Por un problema en el sistema, algunos clientes aparecen más de una vez.

El `customer_id` identifica de manera única a cada cliente.

### Actividades

1. Identifica cuáles son los clientes que aparecen más de una vez.
2. Elimina los registros duplicados manteniendo **la primera aparición** de cada cliente.
3. Repite la operación manteniendo **la última aparición** y compara ambos resultados.

**Investiga:** `drop_duplicates()` y el parámetro que permite decidir qué aparición conservar.

```python
# Creamos el DataFrame con algunos registros duplicados.

customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 102, 104, 103, 105],
    "name": ["Ana", "Carlos", "Maria", "Carlos", "Pedro", "Maria", "Laura"],
    "city": ["Lima", "Arequipa", "Cusco", "Lima", "Tacna", "Arequipa", "Lima"],
    "purchase": [120, 250, 180, 300, 90, 220, 150]
})

customers
```

```python
# 1. Identificamos los registros que tienen un customer_id duplicado.

duplicated_customers = customers[customers["customer_id"].duplicated(keep=False)]

duplicated_customers
```

```python
# 2. Eliminamos duplicados conservando la primera aparición.

customers_first = customers.drop_duplicates(
    subset="customer_id",
    keep="first"
)

customers_first
```

```python
# 3. Eliminamos duplicados conservando la última aparición.

customers_last = customers.drop_duplicates(
    subset="customer_id",
    keep="last"
)

customers_last
```

### Explicación

- `duplicated(keep=False)` permite identificar todas las filas cuyo `customer_id` aparece más de una vez.
- `drop_duplicates()` elimina registros repetidos.
- `keep="first"` conserva la primera aparición.
- `keep="last"` conserva la última aparición.

Observa que para los clientes `102` y `103` cambia la información conservada dependiendo de la opción utilizada.

# 2. Agrupar, transformar el índice y ordenar resultados

Una empresa tiene información sobre las ventas realizadas por diferentes tiendas.

### Actividades

1. Calcula las **ventas totales por tienda**.
2. Convierte el resultado en un DataFrame convencional.
3. Ordena las tiendas desde la que obtuvo mayores ventas hasta la que obtuvo menores ventas e identifica la tienda con mayor facturación.

**Investiga:** `groupby()`, `reset_index()` y `sort_values()`.

```python
# Creamos el DataFrame de ventas.

sales = pd.DataFrame({
    "store": ["A", "B", "A", "C", "B", "C", "A", "B"],
    "product": ["Laptop", "Laptop", "Phone", "Phone",
                "Phone", "Laptop", "Tablet", "Tablet"],
    "sales": [1200, 1500, 800, 950, 1100, 1700, 600, 900]
})

sales
```

```python
# 1. Calculamos las ventas totales por tienda.

sales_by_store = sales.groupby("store")["sales"].sum()

sales_by_store
```

```python
# 2. Convertimos el resultado en un DataFrame convencional.

sales_by_store = sales_by_store.reset_index()

sales_by_store
```

```python
# 3. Ordenamos las tiendas desde la mayor hasta la menor facturación.

sales_by_store = sales_by_store.sort_values(
    by="sales",
    ascending=False
)

sales_by_store
```

### Resultado

La tienda que obtuvo la mayor facturación es la **tienda B**, con **3500** unidades monetarias.

La combinación de `groupby()`, `reset_index()` y `sort_values()` permite pasar de los datos originales a un resumen ordenado y fácil de analizar.

# 3. Construir una tabla resumen con `pivot_table()`

Un restaurante registra la cantidad de pedidos recibidos durante el almuerzo y la cena.

### Actividades

1. Construye una tabla donde las **filas sean los días**, las **columnas sean los tipos de comida (`meal`)** y los valores sean el **total de pedidos**.
2. Determina qué día tuvo más pedidos durante la cena y qué días tuvieron el mayor total de pedidos considerando almuerzo y cena.

**Investiga:** `pivot_table()`.

```python
# Creamos el DataFrame de pedidos.

orders = pd.DataFrame({
    "day": ["Monday", "Monday", "Tuesday", "Tuesday",
            "Wednesday", "Wednesday", "Thursday", "Thursday"],
    "meal": ["Lunch", "Dinner", "Lunch", "Dinner",
             "Lunch", "Dinner", "Lunch", "Dinner"],
    "orders": [35, 50, 40, 65, 30, 70, 45, 60]
})

orders
```

```python
# 1. Construimos la tabla resumen utilizando pivot_table().

orders_pivot = orders.pivot_table(
    index="day",
    columns="meal",
    values="orders",
    aggfunc="sum"
)

orders_pivot
```

```python
# 2. Calculamos el total de pedidos por día.

orders_pivot["Total"] = orders_pivot.sum(axis=1)

orders_pivot.sort_values(by="Total", ascending=False)
```

### Resultado

- El mayor número de pedidos durante la **cena** corresponde a **Wednesday**, con 70 pedidos.
- Los días con mayor cantidad total de pedidos son **Tuesday** y **Thursday**, con 105 pedidos cada uno.

`pivot_table()` permite reorganizar los datos para analizar fácilmente dos dimensiones categóricas.

# 4. Trabajar con fechas y horas

Una empresa de logística registra la fecha y hora de entrega de sus paquetes. Sin embargo, la columna `delivery_time` está almacenada como texto.

### Actividades

1. Convierte `delivery_time` en una columna de tipo **datetime**.
2. Extrae de esa columna el **día de la semana** y la **hora** de cada entrega.
3. Crea una nueva columna llamada `delivery_period` clasificando las entregas como:
   - `Morning`: antes de las 12:00.
   - `Afternoon`: desde las 12:00 hasta antes de las 18:00.
   - `Evening`: desde las 18:00.
4. Cuenta cuántos paquetes fueron entregados en cada período.

**Investiga:** `pd.to_datetime()` y las propiedades `.dt` de Pandas.

```python
# Creamos el DataFrame.
# Observa que delivery_time está almacenado inicialmente como texto.

deliveries = pd.DataFrame({
    "package_id": [1001, 1002, 1003, 1004, 1005],
    "customer": ["Ana", "Carlos", "Maria", "Pedro", "Laura"],
    "delivery_time": [
        "2026-08-20 08:35:00",
        "2026-08-20 14:20:00",
        "2026-08-21 09:15:00",
        "2026-08-21 18:45:00",
        "2026-08-22 11:30:00"
    ]
})

deliveries
```

```python
# 1. Convertimos delivery_time a un tipo datetime.

deliveries["delivery_time"] = pd.to_datetime(
    deliveries["delivery_time"]
)

deliveries.dtypes
```

```python
# 2. Extraemos el día de la semana y la hora.

deliveries["day_of_week"] = deliveries["delivery_time"].dt.day_name()
deliveries["hour"] = deliveries["delivery_time"].dt.hour

deliveries
```

```python
# 3. Clasificamos cada entrega según la hora del día.

def classify_period(hour):
    if hour < 12:
        return "Morning"
    elif hour < 18:
        return "Afternoon"
    else:
        return "Evening"

deliveries["delivery_period"] = deliveries["hour"].apply(classify_period)

deliveries
```

### Resultado

La información final permite analizar la fecha y hora de cada entrega y agruparlas por período:

- **Morning:** 3 paquetes
- **Afternoon:** 1 paquete
- **Evening:** 1 paquete

La propiedad `.dt` permite acceder a diferentes componentes de una columna datetime, como el día, mes, año, hora o día de la semana.
